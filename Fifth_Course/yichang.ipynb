```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def __str__(self):
        return f"Person(name='{self.name}', age={self.age})"

# 使用示例
p = Person("张三", 25)
print(p)           # 输出: Person(name='张三', age=25)
print(str(p))      # 输出: Person(name='张三', age=25)
```

```python
# 示例1：捕获 ValueError
try:
    number = int("abc")
except ValueError as e:
    print(f"转换错误: {e}")
    # e 就是捕获到的 ValueError 异常对象
    # 输出: 转换错误: invalid literal for int() with base 10: 'abc'

# 示例2：捕获多种异常
try:
    result = 10 / 0
except (ZeroDivisionError, ValueError) as e:
    print(f"发生错误: {type(e).__name__} - {e}")
    # 输出: 发生错误: ZeroDivisionError - division by zero

# 示例3：捕获所有异常
try:
    # 一些可能出错的代码
    open("不存在的文件.txt")
except Exception as e:
    print(f"异常类型: {type(e).__name__}")
    print(f"异常信息: {e}")
    # 输出: 
    # 异常类型: FileNotFoundError
    # 异常信息: [Errno 2] No such file or directory: '不存在的文件.txt'
```

```python
# 场景：读取用户输入的文件名，然后读取文件中的数字

try:
    filename = input("请输入文件名：")
    with open(filename, 'r') as f:  # 可能出 FileNotFoundError
        content = f.read()
        number = int(content)       # 可能出 ValueError

except FileNotFoundError:           # 文件找不到错误
    print("❌ 文件不存在！请检查文件名是否正确")
    # 这里可以提示用户重新输入文件名
    
except ValueError:                  # 值转换错误  
    print("❌ 文件内容不是有效的数字！")
    # 这里可以提示用户检查文件内容
    
except PermissionError:             # 权限错误
    print("❌ 没有权限访问这个文件！")
    # 这里可以提示用户检查文件权限
```

```python
try:
    # 代码执行到这里...
    raise ValueError("这是一个错误")  # ← 主动抛出异常
    print("这行不会执行")
    
except ValueError as e:               # ← 接收上面抛出的异常
    print(f"捕获到异常: {e}")         # ← 处理异常
```

```python
def validate_age(age):
    if age < 0:
        raise ValueError("年龄不能为负数")  # 手动抛出
    
    if age > 150:
        raise ValueError("年龄不能超过150岁")  # 手动抛出
        
    return age

# 调用时接收异常
try:
    result = validate_age(-5)
except ValueError as e:  # 接收自己抛出的异常
    print(f"输入无效: {e}")
```

```python
try:
    open("不存在.txt")
except FileNotFoundError as error_obj:
    print(type(FileNotFoundError))     # <class 'type'> - 这是一个类
    print(type(error_obj))            # <class 'FileNotFoundError'> - 这是实例
    print(isinstance(error_obj, FileNotFoundError))  # True - 实例属于这个类
```

```python
class Cat:
    pass

# 1. 实例化一个对象
my_cat = Cat()

# 2. 获取类名称
class_obj = type(my_cat)      # 得到 <class '__main__.Cat'>
print(class_obj)
class_name = class_obj.__name__ # 得到字符串 "Cat"
# class_name2 = <class '__main__.Cat'>.__name__ 简单来说：<class '__main__.Cat'> 是 Python 给你看的一种描述文本，它并不是一种你可以直接写在代码里的语法格式。

name = Cat.__name__  # 合法！输出 "Cat"
print(name)
print(f"对象 my_cat 的类名是: {class_name}")

# 假设你只拿到了 my_cat，不知道它是什么类
c_type = type(my_cat) 

# 因为 c_type 是“真正的类”，你可以直接用它来实例化新的猫！
new_cat = c_type() 
print(type(new_cat).__name__)  # 输出: Cat
```

```python
# ---------------------------------------------------------
# 第一部分：type() 作为“探测器” (查询实例的类)
# ---------------------------------------------------------
class Dog:
    def __init__(self, name):
        self.name = name

my_dog = Dog("旺财")

# 获取类本身（真正的类对象）
dog_class = type(my_dog) 

print("--- 第一部分：探测 ---")
print(f"1. 对象 my_dog 的类型是: {dog_class}") 
print(f"2. 这个类的名字字符串是: {dog_class.__name__}")
print(f"3. 验证它是不是真正的类: {dog_class is Dog}") # 输出 True

# ---------------------------------------------------------
# 第二部分：type() 作为“工厂” (用获取到的类创建新对象)
# ---------------------------------------------------------
# 既然 dog_class 是真正的类，我们就可以像使用 Dog() 一样使用它
another_dog = dog_class("来福")

print("\n--- 第二部分：克隆/创建 ---")
print(f"新狗狗的名字: {another_dog.name}")
print(f"它是 Dog 的实例吗? {isinstance(another_dog, Dog)}")

# ---------------------------------------------------------
# 第三部分：type() 的高级用法 (动态创建类)
# ---------------------------------------------------------
# type(name, bases, dict) 可以在不写 class 关键字的情况下，“凭空”造出一个类
# 语法：type('类名', (父类,), {属性和方法})

def say_hello(self):
    print(f"你好，我是{self.name}！")

# 动态创建一个叫 "Robot" 的类
Robot = type('Robot', (), {'name': '小爱', 'greet': say_hello})

print("\n--- 第三部分：动态造类 ---")
my_robot = Robot()
print(f"动态创建的类名: {type(my_robot).__name__}")
my_robot.greet() # 输出: 你好，我是小爱！
```
