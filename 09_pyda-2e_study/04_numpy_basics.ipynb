```python
import random
import matplotlib.pyplot as plt

# 随机游走步数
nsteps = 1000

# 初始位置
position = 0

# 保存每一步的位置
walk = [position]

for i in range(nsteps):
    step = 1 if random.randint(0, 1) else -1
    position += step
    walk.append(position)

# 画图
plt.figure(figsize=(10, 4))
plt.plot(walk)
plt.title("Random Walk")
plt.xlabel("Step")
plt.ylabel("Position")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

nsteps = 1000

steps = np.random.choice([1, -1], size=nsteps)
walk = steps.cumsum()

plt.figure(figsize=(10, 4))
plt.plot(walk)
plt.title("Random Walk")
plt.xlabel("Step")
plt.ylabel("Position")
plt.show()

np.array([walk.min(),walk.max(),walk.mean(),walk.std()]).round(2)
pd.Series(walk).describe()
```

```python
import numpy as np
import matplotlib.pyplot as plt

nwalks = 5000
nsteps = 1000

# 生成 5000 x 1000 的随机步伐矩阵
# 每个元素都是 1 或 -1
steps = np.random.choice([1, -1], size=(nwalks, nsteps))

# 沿着列方向累加，得到每条随机游走的位置
walks = steps.cumsum(axis=1)

walks.shape

final_positions = walks[:, -1]

summary = {
    "final_min": int(final_positions.min()),
    "final_max": int(final_positions.max()),
    "final_mean": float(final_positions.mean()),
    "final_std": float(final_positions.std()),
    "overall_min": int(walks.min()),
    "overall_max": int(walks.max()),
    "positive_final_ratio": float((final_positions > 0).mean()),
    "hit_30_ratio": float((walks >= 30).any(axis=1).mean()),
    "hit_minus_30_ratio": float((walks <= -30).any(axis=1).mean()),
    "hit_abs_30_ratio": float((np.abs(walks) >= 30).any(axis=1).mean()),
}

summary
```
