```python
# 按照顺序，'小明'填入第一个{}，18填入第二个{}
text = "我叫{}, 今年{}岁。".format("小明", 18)
print(text)
# 输出: 我叫小明, 今年18岁。

# {0} 代表填入 '苹果'，{1} 代表填入 '香蕉'
text = "我喜欢吃{0}，我也喜欢吃{1}，但我最喜欢的还是{0}。".format("苹果", "香蕉")
print(text)
# 输出: 我喜欢吃苹果，我也喜欢吃香蕉，但我最喜欢的还是苹果。

# 乱序填入
text = "亚军是{1}，冠军是{0}。".format("张三", "李四")
print(text)
# 输出: 亚军是李四，冠军是张三。

text = "这台手机的品牌是{brand}，价格是{price}元。".format(brand="苹果", price=5999)
print(text)
# 输出: 这台手机的品牌是苹果，价格是5999元。

# 保留两位小数 (:.2f)
pi = 3.1415926
print("圆周率大约是 {:.2f}".format(pi)) 
# 输出: 圆周率大约是 3.14

# 添加千位分隔符 (:,)
money = 123456789
print("我的存款是 {:,} 元".format(money))
# 输出: 我的存款是 123,456,789 元

# 百分比显示 (:.1%)
ratio = 0.854
print("任务完成率：{:.1%}".format(ratio))
# 输出: 任务完成率：85.4%

# 总结一下： 以前在没有 f-string 的年代，.format() 是全村人的希望。
# 即便现在有了更简洁的 f"我叫{name}" 写法，
# .format() 在很多老代码或者需要重用模板（比如模板在配置文件里，数据在代码里）的场景下，
# 依然发挥着不可替代的作用！

```

```python
pi = 3.14159
print(f"小数: {pi:.2f}")   # 小数: 3.14
print(f"百分比: {pi:.1%}") # 百分比: 314.2%
print(f"科学计数: {pi:e}")  # 科学计数: 3.141590e+00

money = 1234567890
print(f"金额: {money:,}") # 金额: 1,234,567,890
print(f"数量: {money:_}") # 数量: 1_234_567_890


name = "Alice"
# 用竖线 | 包裹起来，方便看清空格的位置
print(f"|{name:<10}|") # |Alice     | (左对齐)
print(f"|{name:>10}|") # |     Alice| (右对齐)
print(f"|{name:^10}|") # |  Alice   | (居中)

user_id = 42
print(f"工号: {user_id:0>5}")  # 工号: 00042 (最常用的补零大法)
print(f"分割线: {'End':-^15}") # 分割线: ------End------


number = 255
print(f"二进制: {number:b}") # 二进制: 11111111
print(f"十六进制: {number:X}") # 十六进制: FF


profit = 500
loss = -200
print(f"今日盈亏: {profit:+}") # 今日盈亏: +500
print(f"今日盈亏: {loss:+}")   # 今日盈亏: -200

money = 12345.6
print(f"最终结果: {money:*>+15.2f}") 
# 最终结果: ******+12,345.60

```

```python
from collections import defaultdict

words = ['apple', 'bat', 'bar', 'atom', 'book']
by_letter = defaultdict(list)          # 自动为新 key 创建空列表
# 当你写 d['a'] 而 'a' 这个键还不存在时，defaultdict 会自动执行 list()，
# 也就是 list() → []，然后把这个空列表作为 'a' 的默认值。

for word in words:
    by_letter[word[0]].append(word)

print(by_letter)
print(dict(by_letter))

print(type(by_letter))                    # <class 'collections.defaultdict'>
print(isinstance(by_letter, dict))
# 输出：defaultdict(<class 'list'>, {'a': ['apple', 'atom'], 'b': ['bat', 'bar', 'book']})
```

```python
words = ['apple', 'bat', 'bar', 'atom', 'book']

# 字典推导式 + 两个 for（外层按首字母，内层再处理）
result = {
    (letter, len(word)): word[::-1]
    for letter in 'ab'                    # 外层
    for word in words if word[0] == letter # 内层
}

print(result)

result = {}
for letter in 'ab':                    # 外层
    for word in words:                 # 内层
        if word[0] == letter:
            result[(letter, len(word))] = word[::-1]
print(result)
```

```python
def my_function(x, y, z=1.5):
    if z > 1:
        return z * (x + y)
    else:
        return z / (x + y)

print(my_function(5, 6))

# 函数可以有一些位置参数（positional）和一些关键字参数（keyword）。
# 关键字参数通常用于指定默认值或可选参数。在上面的函数中，x和y是位置参数，而z则是关键字参数
# 关键字参数必须位于位置参数（如果有的话）之后
```
