```python
from collections import defaultdict

words = ['apple', 'bat', 'bar', 'atom', 'book']
by_letter = defaultdict(list)          # 自动为新 key 创建空列表
# 当你写 d['a'] 而 'a' 这个键还不存在时，defaultdict 会自动执行 list()，
# 也就是 list() → []，然后把这个空列表作为 'a' 的默认值。

for word in words:
    by_letter[word[0]].append(word)

print(by_letter)
print(dict(by_letter))

print(type(by_letter))                    # <class 'collections.defaultdict'>
print(isinstance(by_letter, dict))
# 输出：defaultdict(<class 'list'>, {'a': ['apple', 'atom'], 'b': ['bat', 'bar', 'book']})
```

```python
import re

def clean_strings(strings):
    result = []
    for value in strings:
        value = value.strip() # 删除首尾空格
        value = re.sub('[!#?]', '', value) # 删除标点符号
        value = value.title()
        result.append(value)
    return result

states = ['   Alabama ', 'Georgia!', 'Georgia', 'georgia', 'FlOrIda', 'south   carolina##', 'West virginia?']
clean_strings(states)

# re.sub() 是正则表达式替换函数
# [!#?] 是一个字符类（character class），匹配三个字符中的任意一个：!、# 或 ?
# '' 是替换字符串（空字符串，即删除匹配字符）
# 该行代码会移除 value 中所有 ！、# 和 ? 字符
```

```python
import re

def remove_punctuation(value):
    return re.sub('[!#?]', '', value)

clean_ops = [str.strip, remove_punctuation, str.title]
# str.title 函数将字符串的第一个字符大写，其余字符保持不变
# str.strip 函数将字符串的开头和结尾的空白字符删除
# 当写成 str.strip 而不是 "string".strip() 时，它变成了一个函数对象

def clean_strings(strings, ops):
    result = []
    for value in strings:
        for function in ops:
            value = function(value) 
        result.append(value)
    return result

states = ['   Alabama ', 'Georgia!', 'Georgia', 'georgia', 'FlOrIda', 'south   carolina##', 'West virginia?']
clean_strings(states, clean_ops)
```

```python
def apply_to_list(some_list, f):
    return [f(x) for x in some_list]

ints = [4, 0, 1, 5, 6]
apply_to_list(ints, lambda x: x * 2)

# 本书其余部分一般将其称为lambda函数。
# 它们在数据分析工作中非常方便，因为你会发现很多数据转换函数都以函数作为参数的。
# 直接传入lambda函数比编写完整函数声明要少输入很多字（也更清晰）
```

```python
def add_numbers(x, y):
    return x + y

def add_five(y):
    return add_numbers(5, y)
print(add_five(10))

from functools import partial  # noqa: E402
add_five = partial(add_numbers, 5)
print(add_five(10))
```

```python
# functools.partial 用来 “固定” 函数的部分参数，生成一个新的可调用对象，以后调用时只需要提供剩余的参数。
from functools import partial

def func(a, b, c):
    return (a + b)* c

# 把第一个参数 c 固定为 3W
new_func = partial(func, c = 3)

# 等价于 func(1, b, c)
print(new_func(2, 3))   # 输出 6
```

```python
def power(base, exp):
    return base ** exp

square = partial(power, exp=2)   # 固定指数为 2
print(square(5))                 # 5 ** 2 = 25
```

```python
def count_up_to(n):
    i = 1
    while i <= n:
        yield i          # 暂停并返回 i
        i += 1           # 下次调用从这里继续

# 使用生成器
for num in count_up_to(5):
    print(num, end=' ')
# 输出: 1 2 3 4 5
```

```python
def count_up_to_list(n):
    result = []
    i = 1
    while i <= n:
        result.append(i)
        i += 1
    return result

count_up_to_list(5)
```

```python
def fibonacci():
    a, b = 0, 1
    while True:
        yield a
        a, b = b, a + b

fib = fibonacci()
for _ in range(10):
    print(next(fib), end=" ")
# 输出前 10 个斐波那契数
```

```python
import itertools

logs = [
    "2024-01-01 登录",
    "2024-01-01 购买",
    "2024-01-02 登录",
    "2024-01-02 搜索",
    "2024-01-02 购买",
    "2024-01-03 登录",
]

for date, group in itertools.groupby(logs, key=lambda x: x.split(" ")[0]):
    print(date, list(group))
    actions = [g.split(" ")[1] for g in group]
    print(f"{date}: {actions}")

# 因为 group 是生成器（generator），只能用一次。list(group) 已经把它"榨干"了，后面的 for g in group 就拿不到任何数据了。
```

```python
import itertools

logs = [
    "2024-01-01 登录",
    "2024-01-01 购买",
    "2024-01-02 登录",
    "2024-01-02 搜索",
    "2024-01-02 购买",
    "2024-01-03 登录",
]

for date, group in itertools.groupby(logs, key=lambda x: x.split(" ")[0]):
    group_list = list(group)                     # 只转一次
    print(date, group_list)
    actions = [g.split(" ")[1] for g in group_list]  # 用列表，不是生成器
    print(f"{date}: {actions}")

# groupby 返回的 group 只能消费一次，想用多次就先 list() 存起来。
```
