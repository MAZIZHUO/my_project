```python
with open('examples/ex7.csv') as f:
    lines = list(csv.reader(f))

header, values = lines[0], lines[1:]
data_dict = {h: v for h, v in zip(header, zip(*values))} 
# zip(*values) 将 values 列表中的每个元素（行）解压成单独的列表，然后与 header 列表中的每个元素（列名）配对，创建一个字典，其中键是列名，值是对应列的  数据列表
data_dict
```

```python
import csv

# 1. 先在外面定义好你的“方言”规则
class my_dialect(csv.Dialect):
    lineterminator = '\n'
    delimiter = ','
    quotechar = '"'
    quoting = csv.QUOTE_MINIMAL

# 2. 重新开门，并在这个门关上之前把读取的工作做完
with open('examples/ex7.csv') as f:
    # 让翻译官拿着新手册去读这个刚刚打开的文件 f
    reader = csv.reader(f, dialect=my_dialect)
    
    # 💥重要提示：你要对数据做的任何处理，都必须在这个缩进里完成！
    # 比如你想把数据打印出来看看：
    for line in reader:
        print(line)
        
# 一旦代码退出了上面的缩进跑到这里，文件 f 就又会自动关上了。
```

```python
tables = pd.read_html('examples/fdic_failed_bank_list.html')
failures = tables[0]
```

```python
close_timestamps = pd.to_datetime(failures['Closing Date'])
print(close_timestamps)
close_timestamps.dt.year.value_counts()
#     └─────┘ └────┘ └────────────┘
#        ↓      ↓          ↓
#    datetime  提取年份   统计各年份出现次数
#    Series   (int类型)   (返回Series)
```

```python
# 按月份统计
print(f'{close_timestamps.dt.month.value_counts()}\n')

# 按季度统计
print(close_timestamps.dt.quarter.value_counts(), end='\n\n')

# 按星期几统计
print(close_timestamps.dt.dayofweek.value_counts())

# 组合：按年月统计
print(close_timestamps.dt.to_period('M').value_counts())
```
