```python
def isiterable(obj):
    try:
        iter(obj) # 尝试迭代，iter()函数返回一个迭代器对象，如果对象不可迭代，则会抛出TypeError异常
        return True
    except TypeError: # not iterable
        return False

print(isiterable('a string'))
print(isiterable([1, 2, 3]))
print(isiterable(5))
print(isiterable(5.6))
```

```python
from datetime import datetime, date, time
dt = datetime(2011, 10, 29, 20, 30, 21) # 创建一个datetime对象，表示2011年10月29日20点30分21秒
print(dt.year, dt.month, dt.day, dt.hour, dt.minute, dt.second)

print(dt.date())
print(dt.time())
print(dt.strftime('%m/%d/%Y %H:%M'))
print(dt.strptime('20091031', '%Y%m%d'))

dt.date # datetime对象的date方法，返回一个date对象，包含年月日信息
```

```python
sequence = [1, 2, None, 4, None, 5]
total = 0
for value in sequence:
    if value is None:
        continue
    total += value

print(total)
```

```python
sequence = [1, 2, 0, 4, 6, 5, 2, 1]
total_until_5 = 0
for value in sequence:
    if value == 5:
        break
    total_until_5 += value

print(total_until_5)
```
