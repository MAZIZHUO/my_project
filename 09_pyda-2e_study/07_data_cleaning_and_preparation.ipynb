```python
data = pd.DataFrame(
    {
        'food': ['bacon', 'pulled pork', 'bacon','Pastrami', 'corned beef', 'Bacon','pastrami', 'honey ham', 'nova lox'],
        'ounces': [4, 3, 12, 6, 7.5, 8, 3, 5, 6]
    }
)

meat_to_animal = {
  'bacon': 'pig',
  'pulled pork': 'pig',
  'pastrami': 'cow',
  'corned beef': 'cow',
  'honey ham': 'pig',
  'nova lox': 'salmon'
}
```

```python
lowercased = data['food'].str.lower()
data['animal'] = lowercased.map(meat_to_animal)
# Series.map(dict) 会把 Series 中每个值当作 key 去字典里查，返回对应的 value
data
```

```python
import pandas as pd

s = pd.Series(['cat', 'dog', 'rabbit', 'cat', 'dog'])
# 使用函数转换
result = s.map(lambda x: x.upper()) # 把每个元素转换为大写，lambda 函数是匿名函数，没有名字
print(result)
# 输出:
# 0       CAT
# 1       DOG
# 2    RABBIT
# 3       CAT
# 4       DOG
# dtype: object
```

```python
s = pd.Series(['cat', 'dog', 'rabbit', 'cat', 'dog'])
# 使用字典进行映射
mapping = {'cat': '小猫', 'dog': '小狗', 'rabbit': '兔子'}
result = s.map(mapping)
print(result)
# 输出:
# 0    小猫
# 1    小狗
# 2    兔子
# 3    小猫
# 4    小狗
# dtype: object
```

```python
s = pd.Series(['cat', 'dog', 'rabbit', 'cat', 'dog'])
mapping = {'cat': '小猫', 'dog': '小狗'}  # 没有 rabbit 的映射
result = s.map(mapping)
print(result)
# 输出:
# 0    小猫
# 1    小狗
# 2     NaN  # 找不到映射时变成 NaN
# 3    小猫
# 4    小狗
# dtype: object
```

```python
s = pd.Series(['cat', 'dog', None, 'cat'])
mapping = {'cat': '小猫', 'dog': '小狗'}

# 忽略 NaN 值
result = s.map(mapping, na_action='ignore') 
# na_action='ignore' 的意思是当遇到 NaN 值时，保持 NaN 不变，而不是尝试去映射它。
print(result)
# 输出:
# 0    小猫
# 1    小狗
# 2    None
# 3    小猫
# dtype: object
```

```python
s = pd.Series([1, 2, 3, 1, 2])
mapping = {1: '低', 2: '中', 3: '高'}
result = s.map(mapping)
print(result)
# 输出:
# 0    低
# 1    中
# 2    高
# 3    低
# 4    中
# dtype: object
```

```python
df = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie'],
    'city': ['Beijing', 'Shanghai', 'Beijing'],
    'score': [85, 92, 78]
})

# 对单列使用 map
city_mapping = {'Beijing': '北京', 'Shanghai': '上海'}
df['city_cn'] = df['city'].map(city_mapping)
print(df)
```

```python
s = pd.Series(['a', 'b', 'c'])
mapping_series = pd.Series(['A', 'B', 'C'], index=['a', 'b', 'c'])
result = s.map(mapping_series)
print(result)
# 输出:
# 0    A
# 1    B
# 2    C
# dtype: object
```

```python
df = pd.DataFrame({
    'gender': ['M', 'F', 'Male', 'Female', 'm', 'f']
})

# 统一性别格式
gender_mapping = {
    'M': 'Male', 'm': 'Male', 'Male': 'Male',
    'F': 'Female', 'f': 'Female', 'Female': 'Female'
}
df['gender_clean'] = df['gender'].map(gender_mapping)
print(df)
```

```python
import pandas as pd

data = pd.DataFrame({
    'name': ['Alice', 'Bob'],
    'age': [25, 30],
    'city': ['Beijing', 'Shanghai']
}, index=['row_one', 'row_two'])

print("原始数据:")
print(data)
# 输出:
#           name  age      city
# row_one  Alice   25   Beijing
# row_two    Bob   30  Shanghai

# 修改索引和列名
result = data.rename(index=str.title, columns=str.upper)
print("\n修改后:")
print(result)
# 输出:
#           NAME  AGE      CITY
# Row_One  Alice   25   Beijing
# Row_Two    Bob   30  Shanghai
```

```python
# 假设从 Excel 读取的数据列名不规范
data = pd.DataFrame({
    '  Name ': ['Alice', 'Bob'],
    'AGE': [25, 30],
    'city ': ['Beijing', 'Shanghai']
}, index=['001', '002'])

# 清理列名：去空格 + 转小写
data.columns = data.columns.str.strip().str.lower()

# 或者用 rename
data = data.rename(columns=lambda x: x.strip().lower())

print(data)
# 输出:
#       name  age      city
# 001  Alice   25   Beijing
# 002    Bob   30  Shanghai
```
