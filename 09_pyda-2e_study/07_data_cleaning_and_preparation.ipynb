```python
# ============================================
# 第一部分：导入必要的库
# ============================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 设置中文字体（Windows）
plt.rcParams['font.sans-serif'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

# ============================================
# 第二部分：创建示例数据（实际项目中用 pd.read_csv 等读取）
# ============================================
np.random.seed(42)

# 生成 1000 条销售记录
n = 1000
data = pd.DataFrame({
    '订单编号': [f'ORD{str(i).zfill(5)}' for i in range(1, n+1)],
    '订单日期': pd.date_range('2023-01-01', periods=n, freq='D'),
    '客户ID': np.random.choice([f'C{str(i).zfill(3)}' for i in range(1, 101)], n),
    '产品类别': np.random.choice(['电子产品', '服装', '食品', '家居', '图书'], n, 
                                p=[0.3, 0.25, 0.2, 0.15, 0.1]),
    '产品名称': np.random.choice(['手机', '电脑', 'T恤', '裤子', '零食', '饮料', 
                                 '沙发', '台灯', '小说', '教材'], n),
    '数量': np.random.randint(1, 10, n),
    '单价': np.random.uniform(10, 5000, n).round(2),
    '城市': np.random.choice(['北京', '上海', '广州', '深圳', '杭州', '成都'], n),
    '客户评分': np.random.choice([1, 2, 3, 4, 5], n, p=[0.05, 0.1, 0.2, 0.35, 0.3]),
})

# 计算总金额
data['总金额'] = data['数量'] * data['单价']

# 人为添加一些脏数据（模拟真实场景）
data.loc[np.random.choice(n, 50), '数量'] = np.nan  # 50个缺失值
data.loc[np.random.choice(n, 30), '单价'] = np.nan  # 30个缺失值
data.loc[np.random.choice(n, 20), '城市'] = '未知'   # 20个异常值

print("原始数据形状:", data.shape)
print("\n前5行数据:")
print(data.head())

# ============================================
# 第三部分：数据清洗
# ============================================
print("\n" + "="*60)
print("数据清洗阶段")
print("="*60)

# 1. 查看数据基本信息
print("\n数据信息:")
print(data.info())

print("\n缺失值统计:")
print(data.isnull().sum())

print("\n数值型数据描述统计:")
print(data.describe())

# 2. 处理缺失值
# 数量用中位数填充
data['数量'] = data['数量'].fillna(data['数量'].median())

# 单价用中位数填充
data['单价'] = data['单价'].fillna(data['单价'].median())

# 重新计算总金额
data['总金额'] = data['数量'] * data['单价']

# 3. 处理异常值
# 删除"未知"城市的记录
data = data[data['城市'] != '未知']

# 4. 数据转换
# 提取月份
data['月份'] = data['订单日期'].dt.month
data['星期'] = data['订单日期'].dt.day_name()

# 价格区间分类
data['价格区间'] = pd.cut(data['单价'], 
                          bins=[0, 100, 500, 1000, 5000],
                          labels=['低价(0-100)', '中价(100-500)', '高价(500-1000)', '超高价(1000+)'])

print("\n清洗后数据形状:", data.shape)
print("\n清洗后缺失值统计:")
print(data.isnull().sum())

# ============================================
# 第四部分：探索性数据分析 (EDA)
# ============================================
print("\n" + "="*60)
print("探索性数据分析")
print("="*60)

# 1. 各产品类别销售情况
print("\n各产品类别销售统计:")
category_sales = data.groupby('产品类别').agg({
    '订单编号': 'count',
    '总金额': ['sum', 'mean'],
    '数量': 'sum'
}).round(2)
category_sales.columns = ['订单数', '总销售额', '平均订单金额', '总销量']
print(category_sales)

# 2. 各城市销售情况
print("\n各城市销售统计:")
city_sales = data.groupby('城市')['总金额'].sum().sort_values(ascending=False)
print(city_sales)

# 3. 月度销售趋势
print("\n月度销售趋势:")
monthly_sales = data.groupby('月份')['总金额'].sum()
print(monthly_sales)

# 4. 客户评分分布
print("\n客户评分分布:")
rating_dist = data['客户评分'].value_counts().sort_index()
print(rating_dist)

# 5. 高价值客户分析（消费前10）
print("\n消费Top 10客户:")
top_customers = data.groupby('客户ID')['总金额'].sum().sort_values(ascending=False).head(10)
print(top_customers)

# ============================================
# 第五部分：数据可视化
# ============================================
print("\n" + "="*60)
print("数据可视化")
print("="*60)

# 创建图表
fig, axes = plt.subplots(2, 3, figsize=(18, 12))
fig.suptitle('电商销售数据分析报告', fontsize=16, fontweight='bold')

# 图1：各产品类别销售额
ax1 = axes[0, 0]
category_total = data.groupby('产品类别')['总金额'].sum().sort_values(ascending=False)
category_total.plot(kind='bar', ax=ax1, color='skyblue')
ax1.set_title('各产品类别销售额')
ax1.set_ylabel('销售额（元）')
ax1.tick_params(axis='x', rotation=45)

# 图2：月度销售趋势
ax2 = axes[0, 1]
monthly_sales.plot(kind='line', ax=ax2, marker='o', color='red', linewidth=2)
ax2.set_title('月度销售趋势')
ax2.set_xlabel('月份')
ax2.set_ylabel('销售额（元）')
ax2.grid(True, alpha=0.3)

# 图3：各城市销售额
ax3 = axes[0, 2]
city_sales.plot(kind='barh', ax=ax3, color='lightgreen')
ax3.set_title('各城市销售额')
ax3.set_xlabel('销售额（元）')

# 图4：客户评分分布
ax4 = axes[1, 0]
rating_dist.plot(kind='pie', ax=ax4, autopct='%1.1f%%', colors=['#ff9999', '#66b3ff', '#99ff99', '#ffcc99', '#ff99cc'])
ax4.set_title('客户评分分布')

# 图5：价格区间分布
ax5 = axes[1, 1]
price_dist = data['价格区间'].value_counts()
price_dist.plot(kind='bar', ax=ax5, color='orange')
ax5.set_title('价格区间分布')
ax5.set_ylabel('订单数')
ax5.tick_params(axis='x', rotation=45)

# 图6：销售额与数量散点图
ax6 = axes[1, 2]
sample_data = data.sample(100)  # 抽样100条避免过于密集
ax6.scatter(sample_data['数量'], sample_data['总金额'], alpha=0.6, c='purple')
ax6.set_title('数量与总金额关系')
ax6.set_xlabel('数量')
ax6.set_ylabel('总金额（元）')

plt.tight_layout()
plt.show()

# ============================================
# 第六部分：深入分析
# ============================================
print("\n" + "="*60)
print("深入分析")
print("="*60)

# 1. 产品类别 × 城市 交叉分析
print("\n各城市各类别销售额:")
cross_tab = pd.pivot_table(data, 
                           values='总金额', 
                           index='城市', 
                           columns='产品类别', 
                           aggfunc='sum',
                           fill_value=0)
print(cross_tab.round(2))

# 2. 客户复购分析
customer_orders = data.groupby('客户ID')['订单编号'].nunique()
repeat_customers = customer_orders[customer_orders > 1]
print(f"\n复购客户数: {len(repeat_customers)}")
print(f"复购率: {len(repeat_customers) / customer_orders.nunique() * 100:.2f}%")

# 3. 高价值产品分析
print("\n销售额Top 10产品:")
top_products = data.groupby('产品名称')['总金额'].sum().sort_values(ascending=False).head(10)
print(top_products)

# ============================================
# 第七部分：得出结论和建议
# ============================================
print("\n" + "="*60)
print("分析结论与建议")
print("="*60)

best_category = category_total.idxmax()
best_city = city_sales.idxmax()
best_month = monthly_sales.idxmax()

print(f"""
📊 关键发现：

1. 最畅销产品类别：{best_category}（销售额：{category_total.max():.2f}元）
2. 最佳销售城市：{best_city}（销售额：{city_sales.max():.2f}元）
3. 销售高峰月份：第{best_month}月（销售额：{monthly_sales.max():.2f}元）
4. 平均客户评分：{data['客户评分'].mean():.2f}/5.0
5. 复购客户数：{len(repeat_customers)}人

💡 业务建议：

1. 加大{best_category}类的营销投入
2. 在{best_city}等表现好的城市扩大市场份额
3. 在第{best_month}月前提前备货，抓住销售旺季
4. 关注评分较低的产品，改进质量或服务
5. 设计会员制度，提高客户复购率
""")

# ============================================
# 第八部分：导出结果
# ============================================
# 保存清洗后的数据
# data.to_csv('cleaned_data.csv', index=False, encoding='utf-8-sig')

# 保存分析结果
# category_sales.to_csv('category_sales.csv', encoding='utf-8-sig')
# city_sales.to_csv('city_sales.csv', encoding='utf-8-sig')

print("✅ 分析完成！")
```

```python
import pandas as pd
import numpy as np

# 示例数据
data = pd.DataFrame({
    'A': [1, 2, 4, -5],
    'B': [0.5, 3.5, 1, 2],
    'C': [1, 1, 1, 1]
})

# 步骤分解：
# 1. 计算绝对值
print(np.abs(data))
#    A    B    C
# 0  1  0.5  1.0
# 1  2  3.5  1.0
# 2  4  1.0  1.0
# 3  5  2.0  1.0

# 2. 判断是否大于 3
print(np.abs(data) > 3)
#        A      B      C
# 0  False  False  False
# 1  False   True  False
# 2   True  False  False
# 3   True  False  False

# 3. 按行判断是否有任何一个为 True
print((np.abs(data) > 3).any(axis=1))
# 0    False
# 1     True
# 2     True
# 3     True
# dtype: bool

# 4. 筛选行
result = data[(np.abs(data) > 3).any(axis=1)]
print(result)
#    A    B  C
# 1  2  3.5  1
# 2  4  1.0  1
# 3 -5  2.0  1
```

```python
df = pd.DataFrame(np.arange(5 * 4).reshape((5, 4))) 
# reshape((5, 4))是将一维数组转换为二维数组
# np.arange(5 * 4)生成一个从 0 到 19 的 20 个整数的数组
sampler = np.random.permutation(5)
# permutation(5)生成一个从 0 到 4 的随机排列
# sampler 是一个随机的 5 个整数的数组，每个整数从 0 到 4 且不重复
df.iloc[sampler]
# sampler 中的索引是随机的，所以每次运行结果不同
# .iloc是属性，.take是方法

df.sample(n=3)
# 随机采样 3 行

df.sample(frac=1)  # frac=1 表示返回所有行，但随机排序
```

```python
import pandas as pd

df = pd.DataFrame({
    '姓名': ['张三', '李四', '王五'],
    '城市': ['北京', '上海', '广州'],
    '工资': [8000, 12000, 15000]
})

print(df)
#    姓名  城市     工资
# 0  张三  北京   8000
# 1  李四  上海  12000
# 2  王五  广州  15000

# 对'城市'列进行独热编码
dummies = pd.get_dummies(df['城市'])
print(dummies)
#    北京  广州  上海
# 0   1   0   0
# 1   0   0   1
# 2   0   1   0

# 方法1：直接拼接
df_with_dummies = pd.concat([df, dummies], axis=1) # axis=1 表示按列拼接
print(df_with_dummies)
#    姓名  城市     工资  北京  广州  上海
# 0  张三  北京   8000   1   0   0
# 1  李四  上海  12000   0   0   1
# 2  王五  广州  15000   0   1   0

# 方法2：删除原列
df_final = pd.concat([df.drop('城市', axis=1), dummies], axis=1)
# df.drop('城市', axis=1)删除'城市'列,并返回一个新的DataFrame
print(df_final)
#    姓名     工资  北京  广州  上海
# 0  张三   8000   1   0   0
# 1  李四  12000   0   0   1
# 2  王五  15000   0   1   0
```

```python
text = """Dave dave@google.com
Steve steve@gmail.com
Rob rob@gmail.com
Ryan ryan@yahoo.com
"""
pattern = r'[A-Z0-9._%+-]+@[A-Z0-9.-]+\.[A-Z]{2,4}'

regex = re.compile(pattern, flags=re.IGNORECASE)
# 这行代码用于预编译正则表达式，并设置忽略大小写的标志

print(regex.findall(text))

m = regex.search(text)
print(m)
# search返回的是文本中第一个电子邮件地址（以特殊的匹配项对象形式返回）

for m in regex.finditer(text):
    print(m.group(), m.span())
```

```python
import re

regex = re.compile(r'([A-Z0-9._%+-]+)@([A-Z0-9.-]+\.[A-Z]{2,4})', re.IGNORECASE)
text = "邮箱: admin@company.com"

m = regex.search(text)
# search返回的是文本中第一个电子邮件地址（以特殊的匹配项对象形式返回）

if m:  # 先检查是否匹配成功
    print(f"完整匹配: {m.group()}")       # 'admin@company.com'
    print(f"用户名: {m.group(1)}")        # 'admin'
    print(f"域名: {m.group(2)}")          # 'company.com'
    print(f"位置: {m.span()}")            # (3, 20)
else:
    print("未找到匹配")
```
