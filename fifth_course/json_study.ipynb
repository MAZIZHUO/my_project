```python
from pathlib import Path
import json
json_file = Path("students.json")
contents = json.dumps([
    {"name": "Alice", "age": 18},
    {"name": "Bob", "age": 19},
    {"name": "Charlie", "age": 20}
]) # 创建一个JSON文件,json.dumps()方法将Python对象转换为JSON字符串

json_file.write_text(contents,encoding="utf-8") # 将JSON字符串写入文件

print(f'文件内容已保存到 {json_file}')
```

```python
from pathlib import Path
import json

p = Path('user_info.json')

# 读取用户信息
def read_user_info(path):
    contents = path.read_text(encoding='utf-8')
    user = json.loads(contents)
    print(f'欢迎回来，{user["name"]}')
    print(f'你的年龄为：{user["age"]}')

# 写入用户信息并转化为JSON字符串，JSON字符串中的字符串必须用双引号括起来，布尔类型值会变成小写的true和false
def write_user_info(path):
    name = input('请输入你的名字：')
    age = input('请输入你的年龄：')
    user = {
        'name': name,
        'age': age
    }
    contents = json.dumps(user, ensure_ascii=False) # 将字典转换为JSON字符串,ensure_ascii=False 可以保存中文
    path.write_text(contents, encoding='utf-8')
    print(f'用户信息已保存到{path}中')

if p.exists():
    read_user_info(p)
else:
    write_user_info(p)
```
