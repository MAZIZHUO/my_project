```python
class Singleton:
    _instance = None  # 类变量，保存唯一实例

    def __new__(cls):
        if cls._instance is None:
            print("Creating the singleton instance...")
            cls._instance = super().__new__(cls) # 创建唯一实例，并保存到类变量中
        else:
            print("Returning existing instance...")
        return cls._instance

# 测试
a = Singleton()
b = Singleton()
print(a is b)  # 输出 True！是同一个对象
```

```python
class order:
    def __init__(self,a = 0):
        self.status = a
    
    def __str__(self):
        if self.status == 0:
            return "订单未支付"
        elif self.status == 1:
            return "订单已支付"
        else:
            return "订单状态未知"
    
    def __repr__(self):
        return f"order(status={self.status})"
    
o1 = order()
print(o1)  # 输出：订单未支付
print(repr(o1))  # 输出：order(status=0)

o2 = order(1)
print(o2)  # 输出：订单已支付
print(repr(o2))  # 输出：order(status=1)
```

```python
class order:
    def __init__(self,a = 0):
        self.status = a
    
    def __str__(self):
        if self.status == 0:
            return "订单未支付"
        elif self.status == 1:
            return "订单已支付"
        else:
            return "订单状态未知"
        
    def __call__(self, k):# __call__方法,使得实例对象可以像函数一样被调用
        print(k)

o1 = order()
o1("调用了订单实例")  # 输出：调用了订单实例,说明o1对象被当作函数调用了
        
```

```python
def add(a, b):
    return a + b

class add_class:
    def __init__(self, function):
        self.function = function
    
    def __call__(self, a, b):
        if a > 10 and b > 10:
            return self.function(a, b)
        else:
            print("输入的数字必须都大于10")

add_instance = add_class(add)
a = int(input("请输入数字a："))
b = int(input("请输入数字b："))
add_instance(a, b)  # 根据输入的数字，输出结果或提示
```

```python
def how_much(n, price):
    return n * price

class goods:
    def __init__(self, function, price):
        self.function = function
        self.price = price
    def __call__(self, n):#函数调用,当实例对象被当作函数调用时，执行__call__方法
        return self.function(n, self.price)
    
apple = goods(how_much, 2)  # 每个苹果2元
print(apple(5))  # 买5个苹果，输出10
    
```

```python
names = ["Alice", "Bob", "Charlie"]

class MyList:
    def __init__(self,names):
        self.names = names
    def __len__(self):#当调用len()函数时，执行__len__方法，返回MyList实例的长度
        print("调用了__len__方法")
        return len(self.names)

my_list = MyList(names)
print(len(my_list))  # 输出：3，说明MyList实例的长度是3


```

```python
person = {"name": "Alice",
"age": 30,  "gender": "Female"}

print(person["name"])

class MyDict:
    def __init__(self, data):
        self.data = data
    
    def __getitem__(self, key):#当使用索引访问MyDict实例时，执行__getitem__方法，返回对应的值
        print(f"访问了键：{key}")
        return self.data[key]
    def __setitem__(self, key, value):#当使用索引赋值MyDict实例时，执行__setitem__方法，设置对应的值
        print(f"设置了键：{key}，原值为：{self.data[key]}，修改值为：{value}")
        self.data[key] = value
        return self.data[key]
    def __delitem__(self, key):#当使用del删除MyDict实例的键时，执行__delitem__方法，删除对应的键值对
        print(f"删除了键：{key}，值为：{self.data[key]}")
        del self.data[key]
    def __contains__(self, key):#当使用in关键字检查MyDict实例是否包含某个键时，执行__contains__方法，返回True或False
        print(f"检查了键：{key}，是否存在：{key in self.data}")
        return key in self.data # 返回True或False，表示键是否存在于字典中

my_dict = MyDict(person)
print(my_dict["name"])  # 输出：Alice，说明通过键访问了对应的值
my_dict["age"] = 31  # 设置键为"age"，值为31
print(my_dict["age"])  # 输出：31，说明通过键访问了对应的值

my_dict["gender"] = "Male"
print(my_dict["gender"])

del my_dict["gender"]
print("gender" in my_dict)  # 输出：False
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def __getattribute__(self, attr):
        print(f"访问属性 {attr}")
        return super().__getattribute__(attr)  # 正常返回属性值

p = Person("Alice", 30)
print(p.name)  # 触发 __getattribute__，但这里没有重写，所以直接访问属性
print(p.age)   # 同样触发 __getattribute__，访问age属性
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def __getattr__(self, attr):
        print(f"访问了不存在的属性 {attr}，返回 None")
        return None  # 返回一个默认值，表示属性不存在
    def __getattribute__(self, attr):
        print(f"正在访问属性 {attr}")
        return super().__getattribute__(attr) # 正常访问属性

p = Person("Alice", 30)
print(p.name)  # 输出 Alice，说明正常访问了name属性
print(p.age)   # 输出 30，说明正常访问了age属性
print(p.gender)  # 输出 None，说明访问了不存在的gender属性，返回了None
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def __getattr__(self, attr):
        print(f"访问了不存在的属性 {attr}，返回 None")
        return None  # 返回一个默认值，表示属性不存在
    def __getattribute__(self, attr):
        print(f"正在访问属性 {attr}")
        if attr == "age":
            raise AttributeError("age 被禁止访问")
        return super().__getattribute__(attr)

p = Person("Alice", 30)
print(p.name) 
print(p.age)   
print(p.gender) 
```
