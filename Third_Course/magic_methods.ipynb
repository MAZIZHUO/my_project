```python
# __new__ 方法
class Person:
    def __new__(cls, *args, **kwargs):
        print("Creating a new person...")
        instance = object.__new__(cls)
        return instance

    def __init__(self, name, age):
        print("Initializing a person...")
        self.name = name
        self.age = age

# 创建一个Person对象
person = Person("John", 30)
print(person.name, person.age)
```

```python
class Singleton:
    _instance = None  # 类变量，保存唯一实例

    def __new__(cls):
        if cls._instance is None:
            print("Creating the singleton instance...")
            cls._instance = super().__new__(cls) # 创建唯一实例，并保存到类变量中
        else:
            print("Returning existing instance...")
        return cls._instance

# 测试
a = Singleton()
b = Singleton()
print(a is b)  # 输出 True！是同一个对象
```
