```python
# __new__ 方法
class Person:
    def __new__(cls, *args, **kwargs):
        print("Creating a new person...")
        instance = object.__new__(cls)
        return instance

    def __init__(self, name, age):
        print("Initializing a person...")
        self.name = name
        self.age = age

# 创建一个Person对象
person = Person("John", 30)
print(person.name, person.age)
```

```python
class Singleton:
    _instance = None  # 类变量，保存唯一实例

    def __new__(cls):
        if cls._instance is None:
            print("Creating the singleton instance...")
            cls._instance = super().__new__(cls) # 创建唯一实例，并保存到类变量中
        else:
            print("Returning existing instance...")
        return cls._instance

# 测试
a = Singleton()
b = Singleton()
print(a is b)  # 输出 True！是同一个对象
```

```python
class order:
    def __init__(self,a = 0):
        self.status = a
    
    def __str__(self):
        if self.status == 0:
            return "订单未支付"
        elif self.status == 1:
            return "订单已支付"
        else:
            return "订单状态未知"
    
    def __repr__(self):
        return f"order(status={self.status})"
    
o1 = order()
print(o1)  # 输出：订单未支付
print(repr(o1))  # 输出：order(status=0)

o2 = order(1)
print(o2)  # 输出：订单已支付
print(repr(o2))  # 输出：order(status=1)
```

```python
class order:
    def __init__(self,a = 0):
        self.status = a
    
    def __str__(self):
        if self.status == 0:
            return "订单未支付"
        elif self.status == 1:
            return "订单已支付"
        else:
            return "订单状态未知"
        
    def __call__(self, k):# __call__方法,使得实例对象可以像函数一样被调用
        print(k)

o1 = order()
o1("调用了订单实例")  # 输出：调用了订单实例,说明o1对象被当作函数调用了
        
```

```python
def add(a, b):
    return a + b

class add_class:
    def __init__(self, function):
        self.function = function
    
    def __call__(self, a, b):
        if a > 10 and b > 10:
            return self.function(a, b)
        else:
            print("输入的数字必须都大于10")

add_instance = add_class(add)
a = int(input("请输入数字a："))
b = int(input("请输入数字b："))
add_instance(a, b)  # 根据输入的数字，输出结果或提示
```

```python
def add(a, b):
    return a + b

class Adder:
    def result(self,a,b):
        return add(a,b)

adder_instance = Adder()
print(adder_instance.result(5, 7))  # 输出：12
```

```python
def how_much(n, price):
    return n * price

class goods:
    def __init__(self, function, price):
        self.function = function
        self.price = price
    def __call__(self, n):
        return self.function(n, self.price)
    
apple = goods(how_much, 2)  # 每个苹果2元
print(apple(5))  # 买5个苹果，输出10
    
```
