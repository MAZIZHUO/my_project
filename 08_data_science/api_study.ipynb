```python
import requests

url = "https://api.github.com/search/repositories" # 假设这是你的搜索URL

# 我们把查询参数放在一个字典里
params = {
    "q": "language:python  stars:>=10000",  # 假设你搜索的是 Python 项目
    "sort": "stars",         # 按 Star 数量排序
    "per_page": 100,         # 👑 告诉服务器：请给我每页 100 条数据（这是最大值）
    "page": 1                # 👑 告诉服务器：请给我第 1 页的数据
}

headers = {
    "Accept": "application/vnd.github.v3+json"
}

# 发送请求时，加上 params=params
response = requests.get(url, headers=headers, params=params) # params=params是告诉服务器，请把参数放在 URL 后面。

if response.status_code == 200:
    data = response.json()
    items_list = data['items']
    print(f"总结果数: {data['total_count']}")
    print(f"本次请求返回了 {len(items_list)} 条数据。")  # 这次你会看到 100 条！
```

```python
from numpy import r_
from numpy.random import f
from pytest import mark
import requests
import plotly.express as px

url = "https://api.github.com/search/repositories" # 假设这是你的搜索URL

# 我们把查询参数放在一个字典里
params = {
    "q": "language:python  stars:>=10000",  # 假设你搜索的是 Python 项目
    "sort": "stars",         # 按 Star 数量排序
    "per_page": 100,         # 👑 告诉服务器：请给我每页 100 条数据（这是最大值）
    "page": 1                # 👑 告诉服务器：请给我第 1 页的数据
}

headers = {
    "Accept": "application/vnd.github.v3+json"
}

# 发送请求时，加上 params=params
response = requests.get(url, headers=headers, params=params) 
data = response.json()

repos_links = []
repos_stars = []
hover_texts = []
for repo in data["items"]:
    r_name = repo["name"]
    r_url = repo["html_url"]
    r_link = f" <a href='{r_url}'>{r_name}</a>" # 超链接

    r_stars = repo["stargazers_count"]

    r_owner = repo["owner"]["login"]
    r_desc = repo["description"]
    r_hover = f"{r_name} by {r_owner}<br>{r_desc}"

    repos_links.append(r_link)
    repos_stars.append(r_stars)
    hover_texts.append(r_hover)

fig = px.bar(
    x=repos_links, 
    y=repos_stars, 
    hover_name=hover_texts,

    labels={"x": "Repository", "y": "Stars"},
    title="Most Popular Python Projects on GitHub")
fig.update_traces(marker_color="blue",marker_opacity=0.3)

fig.show()
fig.write_html("popular_python_projects.html")


```

```python
import requests

url = "https://itunes.apple.com/search"

# 定义查询参数
params = {
    "term": "周杰伦",  # 搜索词
    "media": "music",  # 限制为音乐
    "limit": 30         # 只返回 3 条结果
}

# 发送请求，requests 会自动将 params 拼接到 URL 后面
response = requests.get(url, params=params)

if response.status_code == 200:
    data = response.json()
    print(f"共找到 {data['resultCount']} 首符合条件的歌曲：")
    for item in data['results']:
        print(f"歌名: {item['trackName']}, 专辑: {item['collectionName']}")
```

```python
import requests

url = "https://jsonplaceholder.typicode.com/posts"

# 我们想要发布的新文章数据
new_post_data = {
    "title": "我的第一篇 API 文章",
    "body": "这是一篇通过 Python 代码发布的文章！",
    "userId": 1
}

# 发送 POST 请求，将数据以 JSON 格式提交
response = requests.post(url, json=new_post_data)

# POST 请求成功时，状态码通常是 201 (Created)
if response.status_code == 201:
    print("文章发布成功！服务器返回的结果：")
    print(response.json())
else:
    print(f"发布失败，状态码: {response.status_code}")
```

```python
import requests

# DeepSeek 的对话 API 接口地址
url = "https://api.deepseek.com/chat/completions"

# 🔑 替换为你自己在 DeepSeek 开放平台申请的 API Key
api_key = "sk-34a0b4a977784866abe3c74fa066dc3e"

# 1. 设置请求头 (告诉服务器你的身份，以及发送的数据是 JSON 格式)
headers = {
    "Authorization": f"Bearer {api_key}",
    "Content-Type": "application/json"
}

# 2. 设置提交的数据 (按照 API 文档要求的格式)
payload = {
    "model": "deepseek-chat",  # 指定使用的模型，也可以是 "deepseek-reasoner"
    "messages": [
        {"role": "system", "content": "你是一个幽默的助手。"},
        {"role": "user", "content": "给我讲一个关于程序员的冷笑话。"}
    ],
    "temperature": 0.7  # 控制回答的随机性，0-2 之间，值越大越天马行空
}

# 3. 发送 POST 请求
try:
    response = requests.post(url, headers=headers, json=payload, timeout=30)
    response.raise_for_status() # 检查状态码是否是 200 OK

    # 4. 解析服务器返回的 JSON 数据
    data = response.json()
    
    # 从层层嵌套的字典里提取出最终的文本回复
    reply_text = data['choices'][0]['message']['content']
    print("DeepSeek 回复：\n" + reply_text)

except requests.exceptions.RequestException as e:
    print(f"请求发生错误: {e}")
    if response is not None:
        print(f"服务器返回信息: {response.text}")
```
