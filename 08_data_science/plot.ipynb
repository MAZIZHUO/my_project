```python
%config InlineBackend.figure_format = 'retina'

import random
from matplotlib import pyplot as plt

class RandomWalk:
    """一个生成随机漫步数据的类"""

    def __init__(self, num_points=5000):
        """初始化随机漫步的属性"""
        self.num_points = num_points

        # 所有随机漫步都始于(0, 0)
        self.x_values = [0]
        self.y_values = [0]

    def fill_walk(self):
        """计算随机漫步包含的所有点"""

        # 不断漫步，直到列表达到指定的长度
        while len(self.x_values) < self.num_points:

            # 决定前进方向以及沿这个方向前进的距离
            x_direction = random.choice([1, -1])  # 1表示向右，-1表示向左
            x_distance = random.choice([0, 1, 2, 3, 4])  # 前进距离
            x_step = x_direction * x_distance  # 最终的x轴移动距离

            y_direction = random.choice([1, -1])  # 1表示向上，-1表示向下
            y_distance = random.choice([0, 1, 2, 3, 4])  # 前进距离
            y_step = y_direction * y_distance  # 最终的y轴移动距离

            # 拒绝原地踏步
            if x_step == 0 and y_step == 0:
                continue

            # 计算下一个点的x和y值
            next_x = self.x_values[-1] + x_step
            next_y = self.y_values[-1] + y_step

            self.x_values.append(next_x)
            self.y_values.append(next_y)

rw = RandomWalk()
rw.fill_walk()

plt.style.use('seaborn-v0_8')
fig, ax = plt.subplots(figsize=(10, 6),dpi = 600)  # 设置图像大小

point_numbers = range(rw.num_points)
print(point_numbers)

# c=y_values 告诉程序：根据 y 的大小来决定颜色深浅
# cmap=plt.cm.Blues 告诉程序：使用从浅到深的蓝色系
ax.scatter(
    rw.x_values, 
    rw.y_values, 
    c=point_numbers, 
    cmap=plt.cm.Blues, 
    edgecolors='none',
    s=5
    )
ax.set_aspect('equal')  # 设置x轴和y轴的比例相同

ax.scatter(0, 0, c='green', edgecolors='none', s=100)  # 起点，edgecolors='none'去掉点的边框
ax.scatter(rw.x_values[-1], rw.y_values[-1], c='red', edgecolors='none', s=100)  # 终点
ax.axis('off')  # 隐藏坐标轴

plt.savefig('random_walk.png',dpi=600, bbox_inches='tight')  # 保存图像，去除多余的空白,一定要在plt.show()之前调用
plt.show()
```
