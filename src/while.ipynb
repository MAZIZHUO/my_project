```python
n = 9

while n > 0: 
    print(n)
    n -= 1

print('循环结束')
```

```python
msg = ''

while msg != 'q':
    msg = input('请输入：')
    if msg != 'q':
        print(f'你输入的是：{msg}')

print('结束')
```

```python
active = True

while active:
    print("Please tell me what to do:")
    message = input()
    if message == 'quit':
        active = False
    else:
        print("I'll do: " + message)
```
