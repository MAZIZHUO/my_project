```python
def say(name,friends=[]):
    friends.append(name)
    print(friends)

say('xiaoming')
say('xiaohong')
say('xiaoli')#定义函数的时候尽量不要用列表和字典当做默认值
```

```python
def say(name,friends=None):
    if friends is None:#判断friends是否为None
        friends = []
    friends.append(name)
    print(friends)

say('xiaoming')
say('xiaohong')
say('xiaoli')
```

```python
def say(name,friends=None):
    friends = []
    friends.append(name)
    print(friends)

say('xiaoming')
say('xiaohong')
say('xiaoli')
```

```python
def say(name,score1 = 60,score2 = 60):
    print(f'学生{name}的分数是{score1},{score2}')

say(name = '小王',score2 = 90, score1 = 80) # 指定参数,可以改变参数位置
```

```python
x = [1,2,3,4]

def foo(y):
    y[0] = y[0] + 1
    print(y)

foo(x) # y = x,实际参数赋值给形参（引用传值）
print(x) # x被改变，如果x是列表、字典、元组、集合，则改变的是引用，如果x是基本数据类型，则改变的是值
```

```python
x = [1,2,3,4]

def foo(y):
    y[0] = y[0] + 1
    print(y)

foo(x[:]) # 把x复制一份，y = x[:]，这样就不会改变x
print(x)
```

```python
def add(*args):
    sum = 0
    for i in args:
        sum += i
    return sum

add(1,2,3,4,5)
```

```python
def add(*args):
    num = sum(args)
    return num

add(1,2,3,4,5)
```

```python
def add(*args):
    a = sum(args)
    print(a)

print(add(1,2,3,4,5)) #函数没有 return 语句 → 默认返回 None，外层的 print(add(...)) 打印的是函数的返回值 → 即 None
```
