```python
def greet_users(names):
    '''names为形式参数
    '''
    print(f'你好，{names}')

greet_users('小王') # 实际参数，把实际参数传递给形式参数
```

```python
def greet_users(names='zhangsan'):
    '''
    names为形式参数
    zhangsan为默认值

    '''
    print(f'你好，{names}')

greet_users()#默认值
greet_users('mazizhuo')
```

```python
def say(name,score1 = 60,score2 = 60):
    print(f'学生{name}的分数是{score1},{score2}')

say('小王',80,90)
say('小王') # 使用默认参数
```

```python
def say(name,friends=[]):
    friends.append(name)
    print(friends)

say('xiaoming')
say('xiaohong')#定义函数的时候尽量不要用列表和字典当做默认值
```

```python
def say(name,friends=None):
    if friends is None:#判断friends是否为None
        friends = []
    friends.append(name)
    print(friends)

say('xiaoming')
say('xiaohong')
say('xiaoli')
```

```python
def say(name,score1 = 60,score2 = 60):
    print(f'学生{name}的分数是{score1},{score2}')

say(name = '小王',score2 = 90, score1 = 80) # 指定参数,可以改变参数位置
```

```python
x = [1,2,3,4]

def foo(y):
    y[0] = y[0] + 1
    print(y)

foo(x) # y = x,实际参数赋值给形参（引用传值）
print(x) # x被改变，如果x是列表、字典、元组、集合，则改变的是引用，如果x是基本数据类型，则改变的是值
```

```python
x = [1,2,3,4]

def foo(y):
    y[0] = y[0] + 1
    print(y)

foo(x[:]) # 把x复制一份，y = x[:]，这样就不会改变x
print(x)
```
