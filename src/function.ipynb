```python
import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# 1. 核心计算：蒙特卡洛模拟 (与之前逻辑完全一致)
# ==========================================
np.random.seed(42)

num_assets = 5
num_portfolios = 10000
risk_free_rate = 0.02

mean_returns = np.array([0.12, 0.10, 0.15, 0.09, 0.18])
cov_matrix = np.array([
    [ 0.04,  0.01,  0.02, -0.01,  0.03],
    [ 0.01,  0.03,  0.01,  0.00,  0.02],
    [ 0.02,  0.01,  0.06, -0.02,  0.04],
    [-0.01,  0.00, -0.02,  0.02, -0.01],
    [ 0.03,  0.02,  0.04, -0.01,  0.08]
])

weights = np.random.dirichlet(np.ones(num_assets), size=num_portfolios)
port_returns = weights @ mean_returns
port_volatility = np.sqrt(np.einsum('ij,jk,ik->i', weights, cov_matrix, weights))
sharpe_ratios = (port_returns - risk_free_rate) / port_volatility

max_sharpe_idx = np.argmax(sharpe_ratios)
min_vol_idx = np.argmin(port_volatility)

# ==========================================
# 2. 纯 Matplotlib 优雅绘图 (面向对象风格)
# ==========================================

# 创建画布和坐标轴对象，设置背景颜色 (非常浅的高级灰)
fig, ax = plt.subplots(figsize=(12, 8))
fig.patch.set_facecolor('#f8f9fa')  # 图表最外层背景
ax.set_facecolor('#f8f9fa')         # 绘图区域背景

# 手动绘制优雅的底层网格线
ax.grid(True, color='white', linewidth=1.5, zorder=0)

# 绘制散点图 (注意 zorder=2 让散点覆盖在网格线之上，cmap 改用极具热力感的 magma)
scatter = ax.scatter(
    port_volatility, 
    port_returns, 
    c=sharpe_ratios, 
    cmap='magma',        
    marker='o', 
    s=12,                
    alpha=0.7,           
    edgecolors='none',
    zorder=2
)

# 添加颜色条并美化
cbar = fig.colorbar(scatter, ax=ax)
cbar.set_label('Sharpe Ratio', fontsize=12, fontweight='bold', labelpad=15, color='#333333')
cbar.outline.set_visible(False) # 移除颜色条的边框

# 标注极值点
ax.scatter(
    port_volatility[max_sharpe_idx], port_returns[max_sharpe_idx], 
    marker='*', color='#00ff00', s=400, edgecolors='black', linewidth=1,
    label='Max Sharpe Portfolio', zorder=3
)

ax.scatter(
    port_volatility[min_vol_idx], port_returns[min_vol_idx], 
    marker='P', color='#00ccff', s=250, edgecolors='black', linewidth=1,
    label='Min Volatility Portfolio', zorder=3
)

# ==========================================
# 3. 细节修饰 (替代 Seaborn 的 despine)
# ==========================================

# 隐藏顶部和右侧的边框线条
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# 设置左侧和底部边框的颜色和粗细
ax.spines['left'].set_color('#dddddd')
ax.spines['bottom'].set_color('#dddddd')
ax.spines['left'].set_linewidth(1.5)
ax.spines['bottom'].set_linewidth(1.5)

# 设置全英文的标签、标题和刻度颜色
ax.set_title('Efficient Frontier Simulation (Pure Matplotlib)', fontsize=18, fontweight='bold', pad=20, color='#222222')
ax.set_xlabel('Expected Volatility', fontsize=14, fontweight='bold', color='#444444')
ax.set_ylabel('Expected Return', fontsize=14, fontweight='bold', color='#444444')
ax.tick_params(colors='#666666', labelsize=11)

# 设置图例
legend = ax.legend(loc='upper left', fontsize=12, frameon=True, facecolor='white', edgecolor='#eeeeee')
legend.get_frame().set_alpha(0.8)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import polars as pl
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd  # Seaborn 的某些高级功能仍需 pandas DataFrame

# ==========================================
# 1. 全局绘图配置（提升分辨率和字体大小以适应论文）
# ==========================================
# 设置高 DPI 以获得清晰的图像
plt.rcParams['figure.dpi'] = 300 
# 设置 Seaborn 主题，使用白色网格和“muted”调色板
sns.set_theme(style="whitegrid", palette="muted") 

# ==========================================
# 2. 模拟数据生成 (使用 Numpy 和 Polars)
# ==========================================
np.random.seed(101)  # 设置随机种子以保证结果可重现

# 模拟参数
groups = ['Treatment A', 'Treatment B', 'Control']
conditions = ['Pre-test', 'Post-test']
n_samples_per_combo = 50  # 每个组合的样本数

data_list = []
for group in groups:
    for condition in conditions:
        # 根据组别和测试条件模拟不同的正态分布数据
        if group == 'Control':
            mu, sigma = 10, 2
        elif group == 'Treatment A' and condition == 'Pre-test':
            mu, sigma = 12, 2.5
        elif group == 'Treatment A' and condition == 'Post-test':
            mu, sigma = 16, 3
        elif group == 'Treatment B' and condition == 'Pre-test':
            mu, sigma = 11, 2.2
        else: # Treatment B, Post-test
            mu, sigma = 18, 3.5
        
        # 生成正态分布数据
        values = np.random.normal(mu, sigma, n_samples_per_combo)
        
        # 将模拟数据添加到列表中
        for val in values:
            data_list.append({
                'Group': group,
                'Condition': condition,
                'Value': val
            })

# 创建 Polars DataFrame
df = pl.DataFrame(data_list)

# 将数据转换为 Pandas DataFrame，以便调用 Seaborn
pdf = df.to_pandas()

# ==========================================
# 3. 绘制统计图 (使用 Seaborn)
# ==========================================
# 创建图表和坐标轴对象
fig, ax = plt.subplots(figsize=(12, 7))

# --- 步骤 3.1: 绘制箱线图 (展示总体分布) ---
# 使用 hue 参数进行二次分组（Condition）
sns.boxplot(
    data=pdf,
    x='Group',
    y='Value',
    hue='Condition',
    order=groups,        # 强制 X 轴顺序
    hue_order=conditions, # 强制图例顺序
    showfliers=False,    # 不显示箱线图自带的异常点（因为散点图会展示）
    width=0.6,           # 调整箱子宽度
    palette="muted",     # 使用柔和调色板
    ax=ax,
    linewidth=1.2,       # 箱子边框粗细
    boxprops=dict(alpha=.6) # 设置箱子透明度，让底层的网格和上层的散点更清晰
)

# --- 步骤 3.2: 绘制散点图 (展示所有原始数据点) ---
# 使用 dodge=True 让散点按 hue 变量错开，与箱线图对齐
sns.stripplot(
    data=pdf,
    x='Group',
    y='Value',
    hue='Condition',
    order=groups,
    hue_order=conditions,
    size=3,              # 散点大小
    jitter=True,         # 开启抖动，防止数据点重叠
    dodge=True,          # 关键：按 hue 错开
    marker='o',          # 散点形状
    edgecolor='gray',    # 散点边框颜色
    linewidth=0.5,       # 散点边框粗细
    alpha=0.8,           # 散点透明度
    ax=ax,
    palette="muted"      # 使用相同的调色板
)

# ==========================================
# 4. 图表美化与论文格式调整
# ==========================================
# 设置英文标题和坐标轴标签，加粗并调整字号
ax.set_title('Experimental Results: Response Values by Group and Condition', fontsize=18, fontweight='bold', pad=20)
ax.set_xlabel('Experimental Group', fontsize=14, fontweight='bold')
ax.set_ylabel('Measured Value (Arbitrary Units)', fontsize=14, fontweight='bold')

# 调整 X 轴和 Y 轴刻度标签的字号
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)

# 处理图例：因为同时绘制了 boxplot 和 stripplot，默认会出现重复图例
# 我们通过手动获取句柄和标签来清理图例
handles, labels = ax.get_legend_handles_labels()
# 前两个句柄通常来自 boxplot（Patch 对象），后两个来自 stripplot（PathCollection 对象）
# 我们只保留前两个（对应条件的水平），使图例更整洁
ax.legend(handles=handles[:2], labels=labels[:2], title='Condition', title_fontsize=13, fontsize=11, loc='upper left')

# 移除上边框和右边框，使图形更清爽
sns.despine(left=True, bottom=True)

# 调整子图参数以适应标题和标签，防止裁剪
plt.tight_layout()

# ==========================================
# 5. 显示或保存图像
# ==========================================
# 若要保存为图片（如 PNG）供论文使用，取消下行的注释
# plt.savefig('experimental_results_plot.png', dpi=300, bbox_inches='tight')

# 在 Jupyter Notebook 中显示图像
plt.show()
```

```python
# 1. 定义一个“类” (Class) —— 汽车设计图
class Car:
    
    # 这是一个特殊的方法，用于在造车（创建对象）时初始化它的特征
    def __init__(self, brand, color):
        # 3. 定义“属性” (Attributes) —— 汽车的特征
        self.brand = brand  # 品牌属性
        self.color = color  # 颜色属性
        self.is_engine_on = False # 引擎状态属性，默认关闭

    # 4. 定义“方法” (Methods) —— 汽车能做的动作
    def start_engine(self):
        self.is_engine_on = True
        print(f"这辆{self.color}的{self.brand}启动了！")

    def drive(self):
        if self.is_engine_on:
            print(f"这辆{self.brand}正在行驶中...")
        else:
            print("你得先启动引擎！")


# 2. 创建“对象” (Objects) —— 根据图纸造出具体的车
my_car = Car("宝马", "红色")      # 造了一辆红色的宝马（对象 1）
friend_car = Car("特斯拉", "白色") # 造了一辆白色的特斯拉（对象 2）

# 访问对象的“属性”
print(my_car.brand)  # 输出: 宝马
print(friend_car.color) # 输出: 白色

# 调用对象的“方法”
my_car.start_engine() # 输出: 这辆红色的宝马启动了！
my_car.drive()        # 输出: 这辆宝马正在行驶中...

friend_car.drive()    # 输出: 你得先启动引擎！ (因为朋友的车还没调用 start_engine 方法)
```

```python
class Counter:
    # __init__ 就是个保险箱，对象一出生，就把数字存进肚子里
    def __init__(self, start_n):
        self.n = start_n  # 这就是它肚子里的记忆
        
    # 注意，这里的括号里没有 n 了！只有代表自己的 self
    def f(self):
        self.n += 1       # 自己动手修改自己肚子里的记忆
        return self.n

c = Counter(10)
c.f()
c.f()
```
