```python
from datetime import datetime

years = input("请输入你的出生年份： ")
years = int(years)
birthday = input("请输入你是否过完生日\'yes\' or \'no\'： ")

now = datetime.now()
current_year = int(now.strftime("%Y")) 
print(type(current_year))

if birthday == 'yes':
    age = current_year - years
else:
    age = current_year - years - 1

print(f'你今年{age}岁')
```

```python
from datetime import datetime, timedelta
import time

# 1. 获取并格式化当前时间
now = datetime.now()
print(f"当前完整时间: {now}")
print(f"格式化日期: {now.strftime('%Y年%m月%d日')}")
print(f"格式化时间: {now.strftime('%H时%M分%S秒')}")

# 2. 动态年龄计算（替代硬编码的2026年）
birth_year = int(input("请输入出生年份: "))
current_year = now.year  # 直接获取年份属性
age = current_year - birth_year
print(f"你今年 {age} 岁")

# 3. 时间运算示例
one_week_later = now + timedelta(weeks=1)
print(f"一周后的时间: {one_week_later.strftime('%Y-%m-%d %H:%M:%S')}")
```
