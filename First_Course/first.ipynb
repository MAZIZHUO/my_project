```python

def biger_smaller(a):
    biger = []
    smaller = []
    for i in a:
        if i > 5:
            biger.append(i)
        else:
            smaller.append(i)
    return biger, smaller

a = [1,2,3,4,5,6,7]
biger, smaller = biger_smaller(a)

print('dashuzi', biger)
print('xiaoshuzi', smaller)
```

```python
import random

def slot_machine():
    # 定义滚轮上的图案
    symbols = ['🍎', '🍒', '💎', '🍀']
    
    # 随机选三个图案
    slot1 = random.choice(symbols)
    slot2 = random.choice(symbols)
    slot3 = random.choice(symbols)
    
    result = [slot1, slot2, slot3]
    
    # 判断逻辑
    if slot1 == slot2 == slot3:
        if slot1 == '💎':
            message = "特等奖！大富大贵！💰"
        else:
            message = "恭喜中奖！🎉"
        win = True
    else:
        message = "没中奖，再试一次吧。☕"
        win = False
        
    return result, message, win

# 运行游戏
icons, msg, is_win = slot_machine()

print(f"结果是: {' | '.join(icons)}")
print(f"提示: {msg}")
```

```python

#字典
person1 = {
    "name": "John", 
    "age": 36,
    'is marrised': True}

#读取
print(person1["name"])
print(person1.get("name"))
print(person1.get("age"))
#print(person1['gender'])
print(person1.get('gender'))
print(person1.get('gender', 'Male'))

#添加
person1['gender'] = 'Male' 
print(person1)

#删除
del person1['is marrised'] 
print(person1)

#修改
person1['age'] = 40 
print(person1)

#从空集开始
person2 = {}
person2['name'] = 'Mike'
person2['age'] = 30
person2['gender'] = 'Male'
print(person2)

#计数
text = "apple banana apple orange apple"
words = text.split()

count = {}
for word in words:
    count[word] = count.get(word, 0) + 1   # key不存在时默认0，然后+1

print(count)  
# {'apple': 3, 'banana': 1, 'orange': 1}

```

```python
dict1 = {'cc':10,'aa':20,'dd':30,'bb':40}
dict2 = dict1.items()
print(dict2)

#遍历字典
for key,value in dict1.items():
    print(key,value)
    print(f'键：{key} 值：{value}')

#获取字典的键或者值
for key in dict1.keys():
    print(key)

for value in dict1.values():
    print(value)

#字典排序
for key in sorted(dict1.keys()):
    print(key,dict1[key])

for value in sorted(dict1.values()):
    print(value)
```

```python
p1 = {'name':'peter','age':'18','gender':'male'}
p2 = {'name':'joent','age':'19','gender':'female'}
p3 = {'name':'james','age':'20','gender':'male'}
#字典列表
friends = [p1,p2,p3]
print(friends)

for p in friends:
    print(p['name'],p['age'],p['gender'])

# 使用索引
for i in range(len(friends)):
    print(friends[i]['name'],friends[i]['age'],friends[i]['gender'])
```
