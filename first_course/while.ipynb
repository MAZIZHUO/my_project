```python
msg = ''

while msg != 'q':
    msg = input('请输入：')
    if msg != 'q':
        print(f'你输入的是：{msg}')

print('结束')
```

```python
active = True

while active:
    print("Please tell me what to do:")
    message = input()
    if message == 'quit':
        active = False
    else:
        print("I'll do: " + message)
```

```python
value = True

while value:
    name = input("请输入你的名字：")
    if name != 'quit':
        print(f"欢迎来到{name}的机器游戏！,如果想要结束，请输入quit")
    else:
        value = False
        print("游戏结束！")
```

```python
n = 0

while n < 5:
    if n % 2 == 0:
        n += 1
        continue # 跳过当前循环，继续下一次循环
    print(n**2)
    n += 1
```

```python
n = 0

while n < 5:
    if n == 3:
        break # 跳出循环
    print(n)
    n += 1
```

```python
list1 = ['ma', 'zi', 'zhuo']
list2 = []

while list1:
    cur = list1.pop()
    list2.append(cur.title())

print(list(reversed(list2)))
print(list2)
```

```python
users = {}
while True:
    name = input('请输入用户名：')
    age = input('请输入年龄：')
    age = int(age)
    users[name] = age

    if input('是否继续添加用户？(y/n)') == 'n':
        break

print(users)
```
