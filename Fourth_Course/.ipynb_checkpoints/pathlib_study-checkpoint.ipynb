```python
from pathlib import Path

contents = []
p = Path('../Other_Files/names.txt')

while True:
    guest = input("请输入你的名字：,如果要停止，请输入quit：")
    if guest == "quit":
        print("输入结束")
        break
    contents.append(guest)
    p.write_text('\n'.join(contents), encoding='utf-8')
    print(f"{guest} 已被添加到文件")


print(p.read_text(encoding='utf-8'))
```

```python
from pathlib import Path
p = Path('../Other_Files/names.txt')
contents = ''

while True:
    guest = input("请输入你的名字：,如果要停止，请输入quit：")
    if guest == "quit":
        print("输入结束")
        break
    contents += guest + '\n'

p.write_text(contents, encoding='utf-8')
print(p.read_text(encoding='utf-8'))
```
