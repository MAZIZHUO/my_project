```python
x = [3, 1, 4, 1]

# =========================
# ⭐ 一、常用 list 方法（安全执行）
# =========================

print(x.copy())              # 浅拷贝

print(x.count(1))            # 统计元素出现次数
print(x.index(1))            # 查找第一次出现位置

x.append(9)                  # 添加元素（修改原列表）
print(x)

x.extend([7, 8])             # 扩展列表
print(x)

x.insert(2, 100)             # 插入元素
print(x)

print(x.pop())               # 删除最后一个元素并返回
print(x)

x.remove(1)                  # 删除第一个匹配值
print(x)

x.reverse()                  # 原地反转
print(x)

x.sort()                     # 原地排序
print(x)

x.clear()                    # 清空列表
print(x)

# 重新恢复数据用于后续演示
x = [3, 1, 4, 1]

# =========================
# ⭐ 二、运算符对应 dunder（语法糖层）
# =========================

print(x.__add__([5]))       # x + [5]
print(x.__mul__(2))         # x * 2
print(x.__rmul__(2))        # 2 * x

x.__iadd__([6])             # x += [6]（原地修改）
print(x)

print(x.__getitem__(1))     # x[1]

x.__setitem__(1, 999)       # x[1] = 999
print(x)

x.__delitem__(1)            # del x[1]
print(x)

print(3 in x)               # __contains__

print(len(x))               # __len__

# =========================
# ⭐ 三、迭代 / 控制流相关
# =========================

print(list(x.__iter__()))   # for x in x

print(list(reversed(x)))    # reversed(x)

# =========================
# ⭐ 四、比较运算
# =========================

print(x.__eq__([1,2]))      # ==
print(x.__ne__([1,2]))      # !=
print(x.__lt__([9]))        # <
print(x.__gt__([0]))        # >
print(x.__le__([9]))        # <=
print(x.__ge__([0]))        # >=

# =========================
# ⭐ 五、字符串 / 表示方式
# =========================

print(str(x))               # str(x)
print(repr(x))              # repr(x)

# =========================
# ⭐ 六、对象基础信息
# =========================

print(x.__class__)          # 类型
print(x.__doc__)            # 文档
print(x.__sizeof__())       # 内存占用
print(dir(x))               # 所有属性
```
