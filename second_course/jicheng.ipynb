```python
class person:
    def __init__(self, x):
        self.age = x
    
    def show(self):
        print(self.age)

p1 = person(20)
p1.show()

class student(person):
    def __init__(self, x, y):
        super().__init__(x)
        self.grade = y
    
    def show(self):
        super().show()
        print(self.grade)

s1 = student(20, 'A')
s1.show()
```

```python
class person:
    def __init__(self,name):
        self.name = name
    
    def hello(self):
        print('hello,', self.name)

class person2(person):
    def __init__(self, name, age):
        super().__init__(name)
        self.age = age
    
    def nianling(self):
        print('you are', self.age, 'years old')

p2 = person2('Alice', 30)
p2.hello()
p2.nianling()
```

```python
class person: #父类
    def __init__(self, a, b):
        self.name = a
        self.age = b
    
    def show(self):
        print('name:', self.name)
        print('age:', self.age)

class student(person): #子类1
    def __init__(self, a, b, c):
        super().__init__(a, b)
        self.grade = c
    
    def show(self):
        super().show()
        print('grade:', self.grade)

class teacher(person): #子类2
    def __init__(self, a, b, c):
        super().__init__(a, b)
        self.level = c
    
    def show(self):
        super().show()
        print('level:', self.level)

s2 = student('Bob', 22, 'B')
s2.show()
t1 = teacher('Dr. Smith', 45, 'Senior')
t1.show()
```
