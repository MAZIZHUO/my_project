```python
class gailun:
    def q(self):
        print('盖伦技能1')
    def w(self):
        print('盖伦技能2')
    def e(self):
        print('盖伦技能3')
    def r(self):
        print('盖伦技能4')

class ashe:
    def q(self):
        print('艾希技能1')
    def w(self):
        print('艾希技能2')
    def e(self):
        print('艾希技能3')
    def r(self):
        print('艾希技能4')

class player:
    def __init__(self, name, hero):
        self.name = name
        self.hero = hero
    def skill(self):
        self.hero.q()
        self.hero.w()
        self.hero.e()
        self.hero.r()
    def q(self):
        self.hero.q()
    def w(self):
        self.hero.w()

p1 = player('张三', gailun())
p1.q()
p1.w()
p2 = player('李四', ashe()) #这个时候p2的英雄是ashe
p2.q()
```

```python
class students2507:
    def __init__(self,name,techer):
        self.name = name
        self.techer = techer
    def get_teacher(self):
        return self.techer

s1 = students2507('张三','李老师')
s2 = students2507('王五','李老师')
s3 = students2507('赵六','李老师')

print(vars(s1))
print(vars(s2))
print(vars(s3))
print(s1.get_teacher())
```

```python
class students2507:
    teacher = '王老师' #类属性
    def __init__(self,name):
        self.name = name
        self.teacher = '李老师'

    @classmethod #类方法
    def get_teacher(cls):
        return cls.teacher + '你好' #类方法
    
    @staticmethod #静态方法
    def foo():
        return '这是一个普通方法'
    
print(students2507.teacher) #访问类属性
print(students2507.get_teacher()) #访问类方法

p1 = students2507('王小王')
print(p1.name) #访问实例属性
print(p1.teacher) #访问实例属性
print(p1.get_teacher()) #访问类方法
print(p1.foo()) #访问静态方法

```

```python
class person:
    def __init__(self,name,age):
        self.name = name
        self.age = age

p1 = person('张三', 18)
field = input('请输入要访问的属性：')
print(getattr(p1, field)) #getattr()函数可以根据字符串获取对象的属性值
help(getattr)
```

```python
p2 = person('李四', 20)
field = input('请输入要修改的属性：')
new_value = input('请输入新的属性值：')
if field == 'age':
    new_value = int(new_value) #如果修改的是年龄属性，需要将输入的字符串转换为整数
setattr(p2, field, new_value) #setattr()函数可以根据字符串设置
print(p2.age)
```
