```python
from pathlib import Path

contents = []
p = Path('../Other_Files/names.txt')

while True:
    guest = input("请输入你的名字：,如果要停止，请输入quit：")
    if guest == "quit":
        print("输入结束")
        break
    contents.append(guest)
    p.write_text('\n'.join(contents), encoding='utf-8')
    print(f"{guest} 已被添加到文件")


print(p.read_text(encoding='utf-8'))
```

```python
from pathlib import Path
p = Path('../Other_Files/words.txt')
clean_words = []

for w in p.read_text().splitlines():
    clean_words.append(w.split('.')[1].strip())

clean_words = list(set(clean_words)) # 去重
clean_words.sort() # 排序
print(clean_words)
print(len(clean_words))

p = Path('../Other_Files/clean_words.txt')
p.write_text('\n'.join(clean_words), encoding='utf-8')
print(p.read_text())
```

```python
text = "  hello world  "
clean_text = text.strip()
print(repr(clean_text))  # 字符串的 strip() 方法会去除字符串开头和结尾的空白字符,repr() 方法会返回字符串的 repr 表示形式，通常会在字符串周围添加引号，并且会转义特殊字符，这样可以更清晰地看到字符串的内容和格式。

# 包含换行符和制表符
text = "\n\t  hello world  \t\n"
clean_text = text.strip()
print(repr(clean_text))  # 'hello world'
```

```python
# 字符串示例
text = "hello\nworld"

print(str(text))   # hello
                   # world

print(repr(text))  # 'hello\nworld'

# 数字示例  
num = 3.141592653589793

print(str(num))    # 3.141592653589793
print(repr(num))   # 3.141592653589793

# 列表示例
my_list = [1, 2, 'hello']

print(str(my_list))   # [1, 2, 'hello']
print(repr(my_list))  # [1, 2, 'hello']
```

```python
text = "第一行\n第二行\n第三行"
lines = text.splitlines()
print(lines)  # ['第一行', '第二行', '第三行']

# 处理 Windows 和 Unix 换行符
text_windows = "line1\r\nline2\r\nline3"  # Windows 风格
text_unix = "line1\nline2\nline3"         # Unix/Linux/Mac 风格

print(text_windows.splitlines())  # ['line1', 'line2', 'line3']
print(text_unix.splitlines())     # ['line1', 'line2', 'line3']
```
