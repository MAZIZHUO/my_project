```python
class some:
    def __init__(self,x):
        self.n = x
    
    def fool1(self):
        print(self.n)
    
    def fool2(self):
        self.n += 1
        return self.n

s = some(10)
print(s.n)
s.fool1()
print(s.fool2())
s.fool1() # 输出 11
```

```python
class Dog:
    def __init__(self, name):
        self.name = name  # 属性

    def bark(self):       # 方法
        print(f"{self.name} says woof!")

# 创建两个实例（对象）
dog1 = Dog("Buddy")
dog2 = Dog("Max")

print(dog1.name)
dog2.bark()
```

```python
class name:
    def __init__(self, x):
        self.__age = x  # 私有属性
    
    def get_age(self):  # 公共方法
        return self.__age
    
    def set_age(self, age):  # 公共方法
        if age < 0:  # 添加判断条件
            print('年龄不能小于 0')
        else:
            self.__age = age

n = name(20)
print(n.get_age())  # 输出 20
n.set_age(-10)
print(n.get_age()) # 输出 20
n.set_age(10)
print(n.get_age())


#n.__age不能访问，私有属性
```

```python
class name:
    def __init__(self,x):
        self.__age = x
    
    @property # 属性
    def age(self):
        return self.__age
    
    @age.setter
    def age(self,newvalue):
        if newvalue < 0 or newvalue > 120:
            print('年龄必须在 0 到 120 之间')
        else:
            self.__age = newvalue
    
    @age.deleter
    def age(self):
        del self.__age
        print('删除年龄')

name1 = name(18)
print(name1.age)
name1.age = 20
print(name1.age)
name1.age = 200
del name1.age
name1.age = 15
print(name1.age)


```

```python
class name:
    def __init__(self,x):
        self.__age = x

    @property
    def birth_year(self):
        return 2026 - self.__age

mazizhuo = name(27)
print(mazizhuo.birth_year)
#print(mazizhuo.__age) 错误 属性私有

print(mazizhuo)
print(vars(mazizhuo)) # 查看属性
print(type(mazizhuo)) # 查看类型
print(dir(mazizhuo)) # 查看属性和方法
print(mazizhuo.__class__) #等价于 type(mazizhuo)
print(mazizhuo.__dict__) #等价于 vars(mazizhuo)

print(mazizhuo._name__age) # 访问私有属性

mazizhuo._name__age = 99  # 直接覆盖私有属性
print(mazizhuo._name__age)  # 输出 99

mazizhuo.__age = 99 
print(mazizhuo.__age)
print(vars(mazizhuo))
```

```python
x = 123

# 1. bit_length() → 二进制位数
print("bit_length:", x.bit_length())  # 输出: 7 (因为 123 的二进制是 1111011，共 7 位)

# 2. conjugate() → 复数共轭（对整数无变化）
print("conjugate:", x.conjugate())    # 输出: 123

# 3. 属性（不是方法，直接访问）
print("numerator:", x.numerator)      # 输出: 123（分子）
print("denominator:", x.denominator)  # 输出: 1（分母）
print("real:", x.real)                # 输出: 123（实部）
print("imag:", x.imag)                # 输出: 0（虚部）
```

```python
x = 123

# __xxx__ 方法是公开的！
print(x.__add__(456))    # 输出 579（等价于 x + 456）
print(x.__eq__(123))     # 输出 True（等价于 x == 123）

# 但你不应该手动调用它们！
# 正确做法：直接使用 +, == 等运算符
```

```python
# ==========================================
# 第一步：画图纸（定义类）
# 此时内存里只有规则，没有任何实体产品
# ==========================================
class Robot:
    
    # 这是“出厂设置”函数。每当生产一个新机器人时，这个函数会自动运行。
    # self 代表“即将被生产出来的那个机器人自己”
    def __init__(self, name, color):
        self.name = name    # 把参数 name 存进自己的肚子里
        self.color = color  # 把参数 color 存进自己的肚子里
        
    # 这是一个动作（方法）。所有按这张图纸造出来的机器人，都会这个技能。
    def speak(self):
        print(f"滴滴！我是 {self.name}，我的外壳是 {self.color} 的。")

# ==========================================
# 第二步：开动流水线，生产实体（实例化对象）
# ==========================================

# 生产一号机器人，取名叫"瓦力"，涂成"黄色"，然后贴上 robot1 的标签
robot1 = Robot("瓦力", "黄色")

# 生产二号机器人，取名叫"伊娃"，涂成"白色"，然后贴上 robot2 的标签
robot2 = Robot("伊娃", "白色")

# ==========================================
# 第三步：指挥实体做事（调用对象的方法和属性）
# ==========================================

# 1. 让他们各自说话（虽然调用的是同一个图纸里的 speak 方法，但他们记得自己的数据）
robot1.speak()  
robot2.speak()  

# 2. 强行修改某一个实体的内部数据
print("--- 发生意外，瓦力掉进泥坑了 ---")
robot1.color = "黑色"  # 只修改 robot1 的属性
robot1.speak()       # 瓦力变色了
robot2.speak()       # 伊娃依然是白色，不受影响

# 3. 验证他们是内存里两个完全独立的实体
print("--- 身份验证 ---")
print("robot1 和 robot2 是同一个东西吗？", robot1 is robot2)
```
