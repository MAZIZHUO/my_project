```python
def spam(): 
    message = "Spam" 
    word = "spam" 
    for _ in range(100): 
        separator = ", " 
        message += separator + word  
    message += separator 
    message += "spam!" 

    return message 

output = spam()  
print(output)
```

```python
def spam():
    message = "Spam"
    
    # 循环开始
    for i in range(3):
        separator = ", "  # 在循环内部定义的变量
    
    # 循环结束
    # 【证明 1】在循环外部，我们依然可以访问循环内定义的 separator 和 i
    print("循环结束，最后一次的 i 是:", i)
    print("可以在循环外访问 separator:", separator)
    
    return message

# 执行函数
result = spam()

# 【证明 2】在函数外部，无法访问局部变量（取消下面注释会报错）
# print(message)   # 报错！
# print(separator) # 报错！
```
